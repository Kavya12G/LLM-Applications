```python
from google.colab import userdata
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
```

```python
from youtube_transcript_api import YouTubeTranscriptApi

def get_transcript(video_id):
    transcript = YouTubeTranscriptApi().fetch(video_id)

    text = " ".join([item.text for item in transcript])
    return text

# Correct usage
video_id = "ksDAQX6MbeE"
print(get_transcript(video_id))
```

Gemini Setup

```python
from google import genai

client = genai.Client(api_key=GOOGLE_API_KEY)
```

```python
def analyze_video(transcript):
   prompt = f"""
   Analyze this YouTube video transcript and provide:

   1. Summary (short)
   2. Detailed explanation
   3. Key points (bullet list)
   4. Important insights
   5. 5 Questions & Answers

   Transcript:
   {transcript}
   """

   response = client.models.generate_content(
       model="gemini-2.5-flash-lite",
       contents=prompt
   )

   return response.text
```

Run

```python
video_id = "ksDAQX6MbeE"  # replace
transcript = get_transcript(video_id)

result = analyze_video(transcript)
print(result)
```

Chat Model

```python
def ask_question(transcript, question):
   prompt = f"""
   Answer the question based on this video transcript:

   Transcript:
   {transcript}

   Question:
   {question}
   """

   response = client.models.generate_content(
       model="gemini-2.5-flash-lite",
       contents=prompt
   )

   return response.text

# Example
print(ask_question(transcript, "What is the main topic?"))
```
